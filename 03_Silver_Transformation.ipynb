```python
from pyspark.sql.functions import *

BRONZE_TABLE = "workspace.healthcare_demo.bronze_patient_events"
SILVER_TABLE = "workspace.healthcare_demo.silver_patient_events"
SILVER_CHECKPOINT = "/Volumes/workspace/healthcare_demo/checkpoints/silver"
```

```python
bronze_stream = (
    spark.readStream
        .table(BRONZE_TABLE)
)
```

```python
silver_stream = (
    bronze_stream
        .withColumn("event_timestamp", to_timestamp("event_timestamp"))
        .withColumn("patient_age", col("patient_age").cast("int"))
        .withColumn("heart_rate", col("heart_rate").cast("int"))
        .withColumn("spo2", col("spo2").cast("int"))
        .withColumn("temperature", col("temperature").cast("double"))
        .withColumn("systolic_bp", col("systolic_bp").cast("int"))
        .withColumn("diastolic_bp", col("diastolic_bp").cast("int"))
        .withColumn("battery_level", col("battery_level").cast("int"))
        .withColumn("signal_strength", col("signal_strength").cast("int"))
        .withColumn("room_number", col("room_number").cast("int"))
        .withColumn("processed_timestamp", current_timestamp())
)
```

```python
silver_stream = (
    silver_stream
        .filter(col("patient_id").isNotNull())
        .filter(col("event_timestamp").isNotNull())
        .filter(col("heart_rate").isNotNull())
        .filter(col("systolic_bp").isNotNull())
        .filter(col("diastolic_bp").isNotNull())
)
```
