```python
from pyspark.sql.functions import *

SOURCE_PATH = "/Volumes/workspace/healthcare_demo/patient_events"

BRONZE_TABLE = "workspace.healthcare_demo.bronze_patient_events"

CHECKPOINT = "/Volumes/workspace/healthcare_demo/checkpoints/bronze"

SCHEMA_LOCATION = "/Volumes/workspace/healthcare_demo/metadata/bronze"
```

```python
bronze_stream = (
    spark.readStream
        .format("cloudFiles")
        .option("cloudFiles.format", "json")
        .option("cloudFiles.schemaLocation", SCHEMA_LOCATION)
        .load(SOURCE_PATH)
)
```

```python
bronze_stream.printSchema()
```

```python
dbutils.fs.rm("/Volumes/workspace/healthcare_demo/checkpoints/bronze", True)
dbutils.fs.rm("/Volumes/workspace/healthcare_demo/metadata/bronze", True)
```

```python
query = (
    bronze_stream.writeStream
        .format("delta")
        .outputMode("append")
        .option("checkpointLocation", CHECKPOINT)
        .trigger(availableNow=True)
        .toTable(BRONZE_TABLE)
)
```
