```python
# ===============================
# Cell 1 - Imports
# ===============================

import json
import random
import uuid
from datetime import datetime
```

```python
# ===============================
# Cell 2 - Configuration
# ===============================

VOLUME_PATH = "/Volumes/workspace/healthcare_demo/patient_events"

TOTAL_RECORDS = 200000
BATCH_SIZE = 10000

HOSPITALS = [
    "General Hospital",
    "City Medical Center",
    "Regional Healthcare",
    "Community Hospital"
]

WARDS = [
    "ICU",
    "Emergency",
    "Cardiology",
    "Neurology",
    "Pediatrics"
]

GENDERS = [
    "Male",
    "Female"
]

DEVICE_TYPES = [
    "ECG Monitor",
    "Pulse Oximeter",
    "Blood Pressure Monitor",
    "Temperature Sensor"
]
```

```python
def generate_patient_event():

    heart_rate = random.randint(55, 145)
    spo2 = random.randint(85, 100)
    temperature = round(random.uniform(36.0, 40.0), 1)

    return {
        "event_id": str(uuid.uuid4()),
        "event_timestamp": datetime.now(timezone.utc).isoformat(),
        "hospital_id": random.choice(HOSPITALS),
        "ward": random.choice(WARDS),
        "room_number": random.randint(100, 250),
        "patient_id": f"P{random.randint(1000,9999)}",
        "patient_age": random.randint(18, 90),
        "patient_gender": random.choice(GENDERS),
        "device_id": f"D{random.randint(10000,99999)}",
        "device_type": random.choice(DEVICE_TYPES),
        "heart_rate": heart_rate,
        "spo2": spo2,
        "temperature": temperature,
        "systolic_bp": random.randint(100, 160),
        "diastolic_bp": random.randint(60, 100),
        "battery_level": random.randint(20, 100),
        "signal_strength": random.randint(-90, -40)
    }
```

```python
# ===============================
# Cell 4 - Test
# ===============================

event = generate_patient_event()

print(event)
```

```python
dbutils.fs.rm(VOLUME_PATH, True)
dbutils.fs.mkdirs(VOLUME_PATH)
```

```python
# ===============================
# Cell 6 - Generate 200,000 Files
# Each file contains 5 records
# Total Records = 1,000,000
# ===============================

RECORDS_PER_FILE = 5
TOTAL_FILES = 200000

for file_num in range(TOTAL_FILES):

    events = []

    for _ in range(RECORDS_PER_FILE):
        events.append(generate_patient_event())

    file_name = f"{VOLUME_PATH}/event_{file_num:06d}.json"

    json_data = "\n".join(
        json.dumps(event) for event in events
    )

    dbutils.fs.put(
        file_name,
        json_data,
        overwrite=True
    )

    # Progress message every 1000 files
    if file_num % 1000 == 0:
        print(f"Generated {file_num:,} files...")

print(f"\nFinished generating {TOTAL_FILES:,} files.")
print(f"Total Records: {TOTAL_FILES * RECORDS_PER_FILE:,}")
```

```python
# ===============================
# Cell 7 - Verify Files
# ===============================

files = dbutils.fs.ls(VOLUME_PATH)

print(f"Number of files: {len(files)}")

for f in files:
    print(f.name)
```

```python
import json
import random
import uuid
from datetime import datetime, timezone

VOLUME_PATH = "/Volumes/workspace/healthcare_demo/patient_events"

HOSPITALS = [
    "General Hospital",
    "City Medical Center",
    "Regional Healthcare",
    "Community Hospital"
]

WARDS = [
    "ICU",
    "Emergency",
    "Cardiology",
    "Neurology",
    "Pediatrics"
]

GENDERS = [
    "Male",
    "Female"
]

DEVICE_TYPES = [
    "ECG Monitor",
    "Pulse Oximeter",
    "Blood Pressure Monitor",
    "Temperature Sensor"
]
```

```python
# Step 13 - Generate Incremental Test Data

INCREMENTAL_FILES = 100
RECORDS_PER_FILE = 5

for file_num in range(INCREMENTAL_FILES):

    events = [
        generate_patient_event()
        for _ in range(RECORDS_PER_FILE)
    ]

    file_name = (
        f"{VOLUME_PATH}/incremental_{uuid.uuid4()}.json"
    )

    json_data = "\n".join(
        json.dumps(event) for event in events
    )

    dbutils.fs.put(
        file_name,
        json_data,
        overwrite=True
    )

print(f"Generated {INCREMENTAL_FILES} new files.")
print(f"Generated {INCREMENTAL_FILES * RECORDS_PER_FILE} new records.")
```
